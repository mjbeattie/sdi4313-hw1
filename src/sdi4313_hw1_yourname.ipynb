```python
import csv
import json
import requests
from bs4 import BeautifulSoup

def print_table(rows, columns, limit=10):
    """Formats a list of dictionaries into an aligned console table without pandas."""
    if not rows:
        print("No rows to display.")
        return
    display_rows = rows[:limit]
    widths = {
        col: max(len(col), *(len(str(row.get(col, ""))) for row in display_rows))
        for col in columns
    }
    header = " | ".join(col.ljust(widths[col]) for col in columns)
    print(header)
    print("-" * len(header))
    for row in display_rows:
        print(" | ".join(str(row.get(col, "")).ljust(widths[col]) for col in columns))
```

### Q8: Web scraping (10 points)
Visit the website _books.toscrape.com_. This website is intended for scraping practice. You will see that it contains icons of books much like an online bookstore. Look at the HTML for the site (in Chrome, CTRL-click the page and select _View Page Source_). The books on the page are put into HTML articles with the class of _product_pod_. The code below uses a library called _BeautifulSoup_ to scan for the articles and put them into a list called _scraped_books_. It then saves the list as a CSV file to this directory. Run the code and answer the questions in the markdown block below.

```python
scrape_url = "https://books.toscrape.com/"
page_response = requests.get(scrape_url, timeout=30)
page_response.raise_for_status()
page_response.encoding = "utf-8"  # Prevents encoding artifacts on the £ symbol

soup = BeautifulSoup(page_response.text, "html.parser")
book_articles = soup.find_all("article", class_="product_pod")

scraped_books = []
for article in book_articles:
    title = article.find("h3").find("a")["title"].strip()
    price = article.find("p", class_="price_color").get_text(strip=True)
    availability = article.find("p", class_="instock availability").get_text(strip=True)
    
    # Extract the second CSS class from <p class="star-rating Three">
    rating_classes = article.find("p", class_="star-rating").get("class", [])
    rating = rating_classes[1] if len(rating_classes) > 1 else "Unknown"
    
    scraped_books.append({
        "title": title,
        "price": price,
        "availability": availability,
        "rating": rating
    })

book_columns = ["title", "price", "availability", "rating"]

# Export to CSV using csv.DictWriter
with open("scraped_books.csv", mode="w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=book_columns)
    writer.writeheader()
    writer.writerows(scraped_books)
```

### Q9: API Calls (20 points)
The NASDAQ exchanges allows coders to obtain data through API calls without registering keys. You will run the code below to download 100 quotes from the NASDAQ for IBM. The code downloads the quotes and then prints the raw output from the API response. You will see that it returns JSON. To make it useful, it needs to be cleaned up and stored in a CSV file. The code does this for you. Execute the code and answer the questions in the markdown block below.

```python
# Official Nasdaq historical stock quotes endpoint (works for NYSE and Nasdaq tickers)
ticker = "IBM"
api_url = f"https://api.nasdaq.com/api/quote/{ticker}/historical?assetclass=stocks&fromdate=2026-01-01&limit=3500"

# Nasdaq's API requires a standard User-Agent header to accept the request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
    "Accept": "application/json, text/plain, */*"
}

api_response = requests.get(api_url, headers=headers, timeout=30)
api_response.raise_for_status()

payload = api_response.json()
print("Top-Level JSON Keys:", list(payload.keys()))

# Navigate Nasdaq's nested JSON structure to reach the daily rows
raw_rows = payload.get("data", {}).get("tradesTable", {}).get("rows", [])

# Print the raw rows to inspect the API response
print(raw_rows[:10])

# Flatten and clean the JSON rows (removing '$' and ',' formatting) into a chronological list
stock_rows = []
for row in reversed(raw_rows):  # Nasdaq returns newest first; reversed() makes it chronological
    # Convert MM/DD/YYYY to YYYY-MM-DD to match standard ISO format
    m, d, y = row["date"].split("/")
    iso_date = f"{y}-{m}-{d}"
    
    stock_rows.append({
        "date": iso_date,
        "open": row["open"].replace("$", "").strip(),
        "high": row["high"].replace("$", "").strip(),
        "low": row["low"].replace("$", "").strip(),
        "close": row["close"].replace("$", "").strip(),
        "volume": row["volume"].replace(",", "").strip()
    })

stock_columns = ["date", "open", "high", "low", "close", "volume"]
print(f"\nTotal Trading Days Retrieved: {len(stock_rows)}")
print("10 Oldest Trading Days:\n")
print_table(list(stock_rows), stock_columns, limit=10)

# Export full chronological table to CSV
with open("ibm_stock_timeseries.csv", mode="w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=stock_columns)
    writer.writeheader()
    writer.writerows(stock_rows)
```
